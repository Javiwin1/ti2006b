## Research Question
Is there a relationship between students' study hours and their exam scores?

## Data Provenance
The dataset is completely synthetic and was created for this educational exercise. It does not contain real student information.

## Expected Output
A small dataset, descriptive statistics, and a scatter plot showing the relationship between study hours and exam scores.

## Purpose
To demonstrate a reproducible data analysis workflow using VS Code, GitHub, and Google Colab.


```python

import sys
import pandas as pd
import matplotlib.pyplot as plt

print("Python version:", sys.version)
print("Required packages imported successfully!")
```

```python

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [1, 2, 2.5, 3, 4, 5, 6, 7],
    "Exam_Score": [55, 60, 65, 68, 75, 82, 88, 95]
}

df = pd.DataFrame(data)

df
```

```python

print("Dataset Summary:")
print(df.describe())

correlation = df["Study_Hours"].corr(df["Exam_Score"])

print("\nCorrelation:", round(correlation, 3))

plt.figure(figsize=(8, 5))
plt.scatter(df["Study_Hours"], df["Exam_Score"])

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)

plt.show()
```


## Conclusion

The synthetic dataset shows a positive relationship between study hours and exam scores. Students who studied more hours generally achieved higher scores.

The correlation indicates a strong positive association in this example. However, this synthetic dataset does not establish that studying more hours directly causes higher exam scores.
